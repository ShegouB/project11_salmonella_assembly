```python
%%writefile scripts/run_pipeline.sh
#!/usr/bin/env bash
set -euo pipefail

echo "=== Project 11 pipeline started: $(date -Iseconds) ===" | tee -a logs/run_log.txt
echo "Dataset: SRR33931604 (BioSample SAMN48972067, Salmonella enterica, bloodstream infection, Korle Bu Teaching Hospital, Accra, Ghana, 2021-06)" | tee -a logs/run_log.txt

# Step 1: FastQC on raw reads
echo "--- Step 1: FastQC (raw) ---" | tee -a logs/run_log.txt
fastqc data/raw/reads_1.fastq data/raw/reads_2.fastq -o results/fastqc/ 2>&1 | tee -a logs/run_log.txt

# Step 2: TrimmomaticPE
echo "--- Step 2: TrimmomaticPE ---" | tee -a logs/run_log.txt
TrimmomaticPE -threads 4 \
  data/raw/reads_1.fastq data/raw/reads_2.fastq \
  data/trimmed/reads_1_paired.fastq data/trimmed/reads_1_unpaired.fastq \
  data/trimmed/reads_2_paired.fastq data/trimmed/reads_2_unpaired.fastq \
  ILLUMINACLIP:/usr/share/trimmomatic/TruSeq3-PE.fa:2:30:10 \
  LEADING:3 TRAILING:3 SLIDINGWINDOW:4:15 MINLEN:36 2>&1 | tee -a logs/run_log.txt

# Step 2b: FastQC on trimmed reads (before/after comparison)
echo "--- Step 2b: FastQC (trimmed) ---" | tee -a logs/run_log.txt
fastqc data/trimmed/reads_1_paired.fastq data/trimmed/reads_2_paired.fastq -o results/fastqc/ 2>&1 | tee -a logs/run_log.txt

# Step 3: SPAdes assembly (chosen over MEGAHIT per Day 5's accuracy comparison)
echo "--- Step 3: SPAdes assembly ---" | tee -a logs/run_log.txt
spades.py -1 data/trimmed/reads_1_paired.fastq -2 data/trimmed/reads_2_paired.fastq \
  -o results/spades --isolate -t 4 2>&1 | tee -a logs/run_log.txt

echo "=== Pipeline finished: $(date -Iseconds) ===" | tee -a logs/run_log.txt
```

```python
# Extract the FULL 16S sequence — read the whole FASTA record, not just the first line
with open("results/prokka/salmonella_outbreak.ffn") as f:
    lines = f.readlines()

seq_16s = ""
capture = False
for line in lines:
    if line.startswith(">") and "16S ribosomal RNA" in line:
        capture = True
        continue
    elif line.startswith(">") and capture:
        break
    elif capture:
        seq_16s += line.strip()

print(f"Length: {len(seq_16s)} bp")
print(seq_16s)
```

```python
from Bio.Blast import NCBIWWW, NCBIXML

result_handle = NCBIWWW.qblast("blastn", "nt", seq_16s, hitlist_size=5)
blast_record = NCBIXML.read(result_handle)
for alignment in blast_record.alignments[:5]:
    hsp = alignment.hsps[0]
    print(alignment.title)
    print(f"  identity: {hsp.identities}/{hsp.align_length} ({100*hsp.identities/hsp.align_length:.2f}%)")
```
